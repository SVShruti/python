# 1.1 What is Error Handling?

Error handling in Python refers to the process of detecting, managing, and responding to errors (or exceptions) that occur during the execution of a program. An error or exception is an unexpected event that can disrupt the normal flow of a program's execution. Proper error handling allows developers to manage these situations gracefully, either by fixing the issue on the fly, informing the user of the problem, or terminating the program in a controlled manner.

**1.2 Common Types of Errors in Python**

**Syntax Errors:**
    
Syntax errors are caused by incorrect or invalid code structure. These errors are detected by the Python interpreter before the code is executed. For example, missing a colon at the end of a for loop or using incorrect indentation will result in a syntax error.

```python
# Example of Syntax Error
for i in range(5):
    print(i)
```

**Runtime Errors:**

Runtime errors occur during the execution of a program. These errors arise from situations like trying to divide by zero, accessing an undefined variable, or performing operations on incompatible data types.

```python
# Example of Runtime Error

result = 10 / 0  # Raises ZeroDivisionError
```

**Logical Errors:**

Logical errors are not detected by Python's interpreter or runtime. These errors occur due to mistakes in the logic of the program, leading to incorrect or unexpected behavior. For example, using the wrong condition in an if statement.

```python
# Example of Logical Error
def is_even(number):
    return number % 2 == 1  # Incorrect logic for checking even numbers
```

**1.3 Why is Error Handling Important?**

**Error handling is important for several reasons:**

**1. Prevents Crashes:** Proper error handling ensures that a program doesn’t crash unexpectedly when an error occurs.

**2. Improves User Experience:** By handling errors gracefully, users are provided with meaningful error messages or recovery options, instead of cryptic Python error traces.

**3. Aids in Debugging:** Error handling allows developers to identify and understand the sources of errors, making debugging easier.

**4. Enhances Program Reliability:** Programs with robust error handling are more reliable and can deal with unexpected situations better.

# Chapter 2: Basic Error Handling with try and except

**2.1 Basic Structure of try-except**

Error handling in Python is primarily done using the try-except structure. This structure allows you to "try" to execute a block of code and "except" or handle any errors that occur during its execution. The basic structure is as follows:

```python
# Before Try
# Code that runs before the try block

try:
    # Code that may raise an exception
except ExceptionType:
    # Code that runs if the exception occurs
```

This approach ensures that if an error occurs within the try block, the program doesn't crash but instead executes the code within the except block.

# 2.3 Basic Error Handling with try-except

Now that we have an understanding of different types of errors and the Exception class, let’s explore how to handle errors using the try-except structure in Python.

The **try-except block** is the fundamental structure for error handling. It allows you to "try" to execute a block of code, and if an error occurs, to "except" or handle that error gracefully without crashing the program.

The basic structure is:

```python
# Before Try
# Code that runs before the try block

try:
    # Code that may raise an exception
except ExceptionType:
    # Code that runs if the exception occurs
```

The except block only executes if an exception is raised in the try block. If no exception occurs, the except block is skipped.

Let’s explore some scenarios to understand how try-except works in practice.

# 2.4 Example 1: Handling a Division by Zero Error

Handling division by zero is a common use case for error handling since dividing a number by zero in Python raises a ZeroDivisionError.

**Scenario 1: Basic Division Error Handling**

This example demonstrates how to handle a basic division by zero error.

```python
# Before Try
num1 = 10
num2 = 0

try:
    # Try Block
    result = num1 / num2
except ZeroDivisionError:
    # Catch Block
    print("You can't divide by zero!")
```

Explanation: Here, we attempt to divide 10 by 0, which would normally result in a ZeroDivisionError. By catching this error, we prevent the program from crashing and provide a user-friendly message instead.

**Scenario 2: Dynamic Input Division**

This scenario handles division by zero when the numbers are provided by user input.

```python
# Before Try
num1 = int(input("Enter the numerator: "))
num2 = int(input("Enter the denominator: "))

try:
    # Try Block
    result = num1 / num2
except ZeroDivisionError:
    # Catch Block
    print("Division by zero is not allowed!")

print(result)
```

Explanation: The user inputs two numbers, and the program attempts to divide them. If the denominator is zero, the ZeroDivisionError is caught, and a warning is displayed.

**Scenario 3: Division with Conditional Check**

This example shows how you can raise an exception manually before an operation that could lead to a division by zero.



```python
# Before Try
num1 = 25
num2 = 5

try:
    # Try Block
    if num2 == 0:
        raise ZeroDivisionError
    result = num1 / num2
except ZeroDivisionError:
    # Catch Block
    print("Cannot divide by zero!")
```

Explanation: Before performing the division, the program checks if the denominator is zero. If it is, a ZeroDivisionError is manually raised, and the catch block handles it.

# Chapter 3: Advanced Error Handling Techniques

As you become more familiar with error handling in Python, you can take advantage of several advanced techniques that allow for cleaner, more flexible code. These techniques include using the else and finally blocks, manually raising exceptions, and creating custom exceptions.

**3.1 Using else in Error Handling**

The else block is an optional part of the try-except structure. It allows you to execute a block of code only if no exceptions were raised in the try block. This is particularly useful when you want to separate the logic that should execute only if everything goes smoothly from the code that handles exceptions.

**Scenario 1: Division with Else Block**

This scenario demonstrates how to use the else block in a division operation.

```python
# Before Try
num1 = 10
num2 = 2

try:
    # Try Block
    result = num1 / num2
except ZeroDivisionError:
    # Catch Block
    print("You can't divide by zero!")
else:
    # Else Block
    print(f"The result is {result}")
```

Explanation: The else block runs only if the division operation in the try block does not raise an exception. If num2 were zero, the ZeroDivisionError would be caught, and the else block would not execute.

**Scenario 2: File Read with Else Block**

This example uses the else block to read and display the contents of a file only if it is successfully opened.

```python
# Before Try
filename = "example.txt"

try:
    # Try Block
    with open(filename, "r") as file:
        content = file.read()
except FileNotFoundError:
    # Catch Block
    print("File not found.")
else:
    # Else Block
    print(f"File content: {content}")
```

Explanation: The else block runs only if the file is successfully opened and read. If the file doesn't exist, the FileNotFoundError is caught, and the else block is not executed.

**Scenario 3: API Call with Else Block**

This scenario demonstrates how to use an else block with an API call.

```python
# Before Try
import requests

try:
    # Try Block
    response = requests.get("https://jsonplaceholder.typicode.com/todos/1")
    response.raise_for_status()
except requests.exceptions.HTTPError:
    # Catch Block
    print("HTTP error occurred.")
else:
    # Else Block
    print("API call successful.")
    print(response.json())
```

Explanation: If the API call is successful, the else block runs, printing the response data. If an HTTP error occurs, the except block handles it, and the else block is skipped.

**3.2 The finally Block**

The finally block is another optional part of the try-except structure. Code inside the finally block is always executed, regardless of whether an exception was raised or not. This is useful for tasks that need to be performed under all circumstances, such as releasing resources or closing files.

**Scenario 1: File Handling with Finally**

This example demonstrates the use of the finally block in file operations.

```python
# Before Try
filename = "example.txt"

try:
    # Try Block
    file = open(filename, "r")
    content = file.read()
except FileNotFoundError:
    # Catch Block
    print("File not found.")
finally:
    # Finally Block
    file.close()
    print("File has been closed.")
```

Explanation: The finally block ensures that the file is closed whether an exception occurs or not. This prevents resource leaks and is a good practice when working with files.

**Scenario 2: Network Operation with Finally**

This scenario shows how the finally block can be used to clean up after a network operation.

```python
# Before Try
import requests

try:
    # Try Block
    response = requests.get("https://jsonplaceholder.typicode.com/todos/1")
    response.raise_for_status()
    print(response.json())
except requests.exceptions.RequestException as e:
    # Catch Block
    print(f"Network error: {e}")
finally:
    # Finally Block
    print("End of network operation.")
```

Explanation: The finally block runs at the end of the network operation, regardless of whether the request was successful or if an exception occurred. This can be useful for logging or cleaning up resources.

**Scenario 3: Database Connection with Finally**

This example demonstrates using the finally block to ensure that a database connection is always closed.

```python
# Before Try
import sqlite3

try:
    # Try Block
    conn = sqlite3.connect('example.db')
    cursor = conn.cursor()
    cursor.execute("SELECT * FROM example_table")
except sqlite3.DatabaseError as e:
    # Catch Block
    print(f"Database error: {e}")
finally:
    # Finally Block
    conn.close()
    print("Database connection closed.")
```

Explanation: The finally block ensures that the database connection is closed, whether the query is successful or an error occurs. This is crucial for maintaining the stability and performance of database-driven applications.

# 3.3 Raising Exceptions

Sometimes, you may need to manually raise an exception when a certain condition is met in your code. This can be done using the raise keyword. Raising exceptions is useful when you want to enforce certain conditions or signal an error in your code.

**Scenario 1: Raise a Custom Error**

This scenario demonstrates how to manually raise a ValueError if a certain condition is met.

```python
# Before Try
def check_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or above.")
    else:
        print("Age is valid.")

try:
    # Try Block
    check_age(16)
except ValueError as e:
    # Catch Block
    print(e)
```

Explanation: The check_age function raises a ValueError if the age provided is less than 18. This error is then caught in the try-except block, and an appropriate message is printed.

**Scenario 2: Raise a TypeError**

This example shows how to raise a TypeError if the input arguments are not of the expected type.

```python
# Before Try
def add_numbers(a, b):
    if not isinstance(a, int) or not isinstance(b, int):
        raise TypeError("Both arguments must be integers.")
    return a + b

try:
    # Try Block
    result = add_numbers(10, "20")
except TypeError as e:
    # Catch Block
    print(e)
```

Explanation: The add_numbers function checks if both inputs are integers. If not, it raises a TypeError, which is then caught and handled in the try-except block.

**Scenario 3: Raise a Custom Exception in a Class**

This scenario demonstrates how to create and raise a custom exception.

```python
# Before Try
class NegativeNumberError(Exception):
    pass

def check_positive(number):
    if number < 0:
        raise NegativeNumberError("Negative number is not allowed.")
    return number

try:
    # Try Block
    check_positive(-10)
except NegativeNumberError as e:
    # Catch Block
    print(e)
```

Explanation: A custom exception NegativeNumberError is created by inheriting from the Exception class. The check_positive function raises this custom exception if a negative number is passed. The exception is then caught and handled in the try-except block.

# 3.4 Custom Exceptions

Creating custom exceptions allows you to define specific error conditions relevant to your application. Custom exceptions are created by subclassing the Exception class or one of its subclasses.

**Scenario 1: Custom Error for Negative Numbers**

This scenario shows how to define and raise a custom exception for negative numbers.

```python
# Before Try
class NegativeNumberError(Exception):
    pass

def check_positive(number):
    if number < 0:
        raise NegativeNumberError("Negative number is not allowed.")
    return number

try:
    # Try Block
    check_positive(-5)
except NegativeNumberError as e:
    # Catch Block
    print(e)
```

Explanation: The NegativeNumberError exception is defined to handle cases where a negative number is not allowed. The check_positive function raises this exception, and it is caught in the try-except block.

**Scenario 2: Custom Exception for Invalid Input**

This scenario shows how to create a custom exception for invalid input.

```python
# Before Try
class InvalidInputError(Exception):
    pass

def validate_input(value):
    if not isinstance(value, str):
        raise InvalidInputError("Input must be a string.")
    return value

try:
    # Try Block
    validate_input(123)
except InvalidInputError as e:
    # Catch Block
    print(e)
```

Explanation: The InvalidInputError exception is created to handle cases where the input is not a string. The validate_input function raises this exception when the input is of the wrong type, and the exception is caught and handled in the try-except block.

**Scenario 3: Custom Exception in a File Operation**

This scenario demonstrates the creation of a custom exception to handle file size constraints.

```python
# Before Try
class FileSizeError(Exception):
    pass

def check_file_size(file_path):
    import os
    if os.path.getsize(file_path) > 1000000:  # 1MB size limit
        raise FileSizeError("File size exceeds limit.")

try:
    # Try Block
    check_file_size("largefile.txt")
except FileSizeError as e:
    # Catch Block
    print(e)
```

Explanation: The FileSizeError exception is defined to handle cases where a file exceeds a specified size limit. The check_file_size function raises this exception if the file is too large, and it is caught and handled in the try-except block.

# Chapter 4: Common Scenarios and Best Practices
In this chapter, we'll explore some common scenarios where error handling is essential, along with best practices for writing clean and maintainable code.

**4.1 Handling File I/O Errors**

File operations are prone to errors, such as trying to open a non-existent file or writing to a read-only file. Proper error handling ensures that your program can deal with these issues gracefully.

**Scenario 1: Reading a Non-Existent File**

```python
# Before Try
filename = "nonexistentfile.txt"

try:
    # Try Block
    with open(filename, "r") as file:
        content = file.read()
except FileNotFoundError:
    # Catch Block
    print(f"The file {filename} was not found.")
```

Explanation: This example attempts to read a file that does not exist. The FileNotFoundError is caught, and a friendly error message is displayed.

**Scenario 2: Writing to a Read-Only File**

```python
# Before Try
filename = "readonlyfile.txt"

try:
    # Try Block
    with open(filename, "w") as file:
        file.write("Trying to write to a read-only file.")
except IOError:
    # Catch Block
    print(f"Cannot write to the file {filename}. It might be read-only.")
```

Explanation: This example tries to write to a file that may be read-only. The IOError is caught, and an appropriate error message is displayed.

**Scenario 3: Ensuring File Closure with Finally**

```python
# Before Try
filename = "example.txt"

try:
    # Try Block
    file = open(filename, "r")
    content = file.read()
except FileNotFoundError:
    # Catch Block
    print(f"The file {filename} was not found.")
finally:
    # Finally Block
    file.close()
    print("File has been closed.")
```

Explanation: The finally block ensures that the file is closed, regardless of whether an exception occurred during the file operation.

**4.2 Handling User Input Errors**

User input is often unpredictable, and handling potential errors is crucial for creating robust applications.

**Scenario 1: Validating Integer Input**

```python
# Before Try
try:
    # Try Block
    number = int(input("Enter a number: "))
except ValueError:
    # Catch Block
    print("Please enter a valid integer.")
```

Explanation: This example prompts the user to enter a number. If the input is not a valid integer, a ValueError is caught, and an error message is displayed.

**Scenario 2: Ensuring Positive Input**

```python
# Before Try
try:
    # Try Block
    number = int(input("Enter a positive number: "))
    if number < 0:
        raise ValueError("The number must be positive.")
except ValueError as e:
    # Catch Block
    print(e)
```

Explanation: This example ensures that the user inputs a positive integer. If the input is negative or invalid, the ValueError is caught, and the error message is displayed.

**Scenario 3: Handling Non-Numeric Input**

```python
# Before Try
try:
    # Try Block
    value = input("Enter a numeric value: ")
    if not value.isdigit():
        raise ValueError("The input must be a numeric value.")
except ValueError as e:
    # Catch Block
    print(e)
```
